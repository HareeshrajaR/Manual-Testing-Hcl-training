### 1.	Print all prime numbers between input range (Ex – input 20 50, prints all prime numbers between 20 and 50).


```python
input1=int(input())
input2=int(input())
for i in range(input1,input2+1):
  count=0
  for j in range(1,i+1):
    if(i%j==0):
      count+=1
  if(count==2):
    print(i,end=" ")
```

### 2.	Factorial using recursion


```python
def fact_num(n):
  if(n==0):
    return 1
  return n*fact_num(n-1)
n=int(input())
print(fact_num(n))
```

### 3.	Square of numbers using lambda

```python
square=lambda a:a**2
a=int(input())
print(square(a))
```

### 4.	Find the second largest element in a list

```python
arr=list(map(int,input().split()))
arr.sort()
print(arr)
print(arr[len(arr)-2])
```

### 5.	Count frequency of characters in a string

```python
str1=input()
str=str1.lower()
dict={}
for i in str:
  if i in dict:
    dict[i]+=1
  else:
    dict[i]=1
print(dict)
```

### 6.	Calculate area of a circle using math library

### 7.	Reverse a string without using built‑in reverse

```python
str=input()
n=len(str)
res=""
for i in range(-1,-n-1,-1):
  res+=str[i]
print(res)
```

### 8.	Remove duplicates from a list

```python
arr=list(map(int,input().split()))
set_fun=set()
res=[]
for i in arr:
  set_fun.add(i)
for i in set_fun:
  res.append(i)
print(res)
```

### 10.	Fibonacci series using recursion


```python
def fibonnaci(a,b,n):
  if(n==0):
    return 0
  print(a,end=" ")

  fibonnaci(b,a+b,n-1)
fibonnaci(0,1,7)
```
